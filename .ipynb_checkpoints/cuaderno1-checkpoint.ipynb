```python

from faker import Faker
import pandas as pd
fake = Faker()
```

```python
lista = []
```

```python
for i in range(0,10000):
    lista.append({'nombre': fake.name(), 'apellido': fake.last_name(), 'email': fake.email(), 'direccion': fake.address()})
```

```python
#print(lista)
```

```python
data = pd.DataFrame(lista)
```

```python
data
```

```python
import random
```

```python
def age(params):
    ages = ("30", "40")
    return random.choice(ages)
    
```

```python
def province(params):
    provinces = ('Loja', 'Pichincha', 'Guayas', 'Azuay', 'Manabí')
    return random.choice(provinces)
```

```python
data['age'] = data['nombre'].apply(age)
data['province'] = data['nombre'].apply(province)
```

```python
data
```

```python
data.to_csv('Data/mdata.csv', index=False)
```
